**№1.** (1 point) Найдите субдифференциалы для следующих функций:
- $f(x) = \max(0, 1 − ax), \ \ a \ — \ const$,  во всех точках

- $f(x) = \sin x, \ x \in [0; \frac{3}{2} \pi]$

**№1. - Решение**: 
- Рассмотрим 3 случая: $a = 0$, $a > 0$, $a < 0$. 
    - $a = 0$: $\partial f(x) = df(x) = 0$ - исходная функция просто всюду диф-емая прямая
    - $a > 0$: $f(x) = \begin{cases}1 - ax, x \in \left(-\infty; \frac{1}{a}\right)\\ 0, x \in \left[\frac{1}{a}; +\infty\right)\end{cases}$, 
    $\partial f(x) = \begin{cases}-a, x \in \left(-\infty; \frac{1}{a}\right)\\ k, k \in \left[-a;0\right], x = \frac{1}{a}, \\ 0, x \in \left(\frac{1}{a}; +\infty\right)\end{cases}$
    - $a < 0$: $f(x) = \begin{cases}0, x \in \left(-\infty; \frac{1}{a}\right)\\ 1 - ax, x \in \left[\frac{1}{a}; +\infty\right)\end{cases}$, 
    $\partial f(x) = \begin{cases}0, \left(-\infty; \frac{1}{a}\right) \\ t, t \in \left[a;0\right], x = \frac{1}{a} \\ -a, x \in \left(\frac{1}{a}; +\infty\right)\end{cases}$

- $f(x)$ гладкая, поэтому $df(x) = \partial f(x) \forall x \in X$, т.о. $\partial f(x) = \cos x, x \in [0; \frac{3}{2} \pi]$

**№2.** (1 point) Подробнее ознакомьтесь с материалом по ROC-AUC по [ссылке](https://alexanderdyakonov.wordpress.com/2017/07/28/auc-roc-%D0%BF%D0%BB%D0%BE%D1%89%D0%B0%D0%B4%D1%8C-%D0%BF%D0%BE%D0%B4-%D0%BA%D1%80%D0%B8%D0%B2%D0%BE%D0%B9-%D0%BE%D1%88%D0%B8%D0%B1%D0%BE%D0%BA/) и следующую решите задачу:
  - Пусть на ответах алгоритма $m$ (принимающих значения от $0$ до $1$) задано распределение объектов класса $1$ (доля объектов класса $1$ в зависимости от ответа алгоритма) следующим образом:
    - $\mathbb{P} (m \in [a, b] \ | \ y = 1) = \int_a^b p(z)dz$

  - Распределение объектов класса 0 задаётся так:
    - $\mathbb{P} (m \in [a, b] \ | \ y = 0) = \int_a^b (2 − p(z))dz$

где $p(z) = −1.5z^2 + 3z$

Найдите вероятностные оценки на величины TPR, FPR и ROC-AUC.

Пусть $t \in [a,b] \subseteq [0,1]$ - пороговое значение для определения элементов класса 1 (Positive class). Тогда по определению $TRP = \mathbb{P}\left(m \in \left[t, b\right] \big| y = 1\right)$, $FPR = \mathbb{P}\left(m \in \left[t, b\right] \big| y = 0\right)$. И $ROC - AUC = \int\limits_{\omega} TPR d FPR$. Подставив значения из условия, получим:

$TPR = \int\limits_t^b (-1.5z^2 + 3 z) dz = 1.5 b^2 - 0.5 b^3 - 1.5 t^2 + 0.5 t^3 \big|_{b=1} = 1 - 1.5 t^2 + 0.5 t^3$

$FPR = \int\limits_t^b (2 + 1.5z^2 - 3 z) dz = 2 b - 1.5 b^2 + 0.5 b^3 - 2 t + 1.5 t^2 - 0.5 t^3 \big|_{b=1} = 1 - 2 t + 1.5 t^2 - 0.5 t^3$

$ROC-AUC = \int\limits_{\omega} (1 - 1.5 t^2 + 0.5 t^3) d(1 - 2 t + 1.5 t^2 - 0.5 t^3)$ $= \int\limits_{0}^1(1 - 1.5 t^2 + 0.5 t^3) (-2 + 3 t - 1.5 t^2)dt = -0.75$

```python
import pandas as pd
import nltk
import numpy as np
import sklearn
nltk.download('wordnet')
nltk.download('stopwords')
```

```python
df = pd.read_csv("./hw2_data.csv")
df.head(10)
```

```python
def lemmatize(splitted_str: list[str]) -> list[str]:
    wnl = nltk.stem.WordNetLemmatizer()
    s = []
    for el in splitted_str:
        # s = f"{s} {wnl.lemmatize(el)}"
        s.append(wnl.lemmatize(el))
    return s

def filter(lemmatized_str: list[str]) -> str:
    from nltk.corpus import stopwords
    eng_stopwords = set(stopwords.words('english'))
    s = ""
    for w in lemmatized_str:
        if w not in eng_stopwords:
            s = f"{s} {w}"
    return s
```

```python
df['text'] = df.text.str.lower().str.replace(r'[^a-zа-я0-9]', ' ', regex=True).apply(
    lambda x: filter(lemmatize(x.split()))
)
```
